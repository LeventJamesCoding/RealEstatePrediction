```python
import pandas as pd
from google.colab import userdata

dataset_url = userdata.get('DATASET_URL')
df = pd.read_csv(dataset_url)
display(df.head(5))
```

```python
pd.set_option('display.max_columns', None)
df.head(20)
```

```python
df_cleaned = df.drop(['Order', 'PID'], axis=1)
print(df_cleaned.to_string())
#Target attribute = SalePrice
```

```python
import numpy as np
from sklearn.model_selection import KFold, cross_validate, cross_val_predict
from sklearn.linear_model import Lasso
from sklearn.ensemble import RandomForestRegressor
from xgboost import XGBRegressor
from sklearn.metrics import mean_squared_error, r2_score



#PREPROCESSING
cols_to_fill_none = ['Pool QC', 'Misc Feature', 'Alley', 'Fence', 'Fireplace Qu', 'Garage Type', 'Garage Finish', 'Garage Qual', 'Garage Cond', 'Bsmt Qual', 'Bsmt Cond',
                     'Bsmt Exposure', 'BsmtFin Type 1', 'BsmtFin Type 2', 'Mas Vnr Type']

for col in cols_to_fill_none:
    if col in df_cleaned.columns:
        df_cleaned[col] = df_cleaned[col].fillna("None")

numerical_cols = df_cleaned.select_dtypes(include=['int64', 'float64']).columns
for col in numerical_cols:
    if col != 'SalePrice':
        df_cleaned[col] = df_cleaned[col].fillna(0)

categorical_cols = df_cleaned.select_dtypes(include=['object']).columns
for col in categorical_cols:
    if df_cleaned[col].isnull().sum() > 0:
        df_cleaned[col] = df_cleaned[col].fillna(df_cleaned[col].mode()[0])
```

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# 1. Missing Values Analysis & Visualization
missing_data = df.isnull().sum()
missing_data = missing_data[missing_data > 0].sort_values(ascending=False)
missing_percent = (missing_data / len(df)) * 100

missing_df = pd.DataFrame({
    'Missing Count': missing_data,
    'Missing Percent (%)': missing_percent
})

print("--- Top 15 Features with Missing Values ---")
print(missing_df.head(15))

# Plotting the missing data
plt.figure(figsize=(14, 6))
sns.barplot(x=missing_percent.index, y=missing_percent.values, palette='viridis')
plt.xticks(rotation=90)
plt.xlabel('Features', fontsize=12)
plt.ylabel('Percentage of Missing Values (%)', fontsize=12)
plt.title('Missing Data Percentage by Feature', fontsize=15, fontweight='bold')
plt.tight_layout()
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from scipy import stats

# Select the most critical skewed features to showcase in the report
# You can change these column names based on your dataset's exact spelling
features_to_showcase = ['SalePrice', 'Lot Area', 'Mas Vnr Area', 'Open Porch SF']

for feature in features_to_showcase:
    # Ensure the feature exists in the dataframe to avoid errors
    if feature not in df.columns:
        print(f"Feature '{feature}' not found in dataframe. Skipping.")
        continue

    # Drop NaNs for accurate statistical calculation
    original_data = df[feature].dropna()

    # Create a figure with 2 side-by-side plots
    fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(16, 6))

    # ==========================================
    # PLOT 1: ORIGINAL DISTRIBUTION (BEFORE)
    # ==========================================
    sns.histplot(original_data, kde=True, stat='density', ax=ax1, color='lightcoral', bins=40)

    # Fit a theoretical Normal Distribution curve (Black dashed line)
    mu, std = stats.norm.fit(original_data)
    xmin, xmax = ax1.get_xlim()
    x = np.linspace(xmin, xmax, 100)
    p = stats.norm.pdf(x, mu, std)
    ax1.plot(x, p, 'k--', linewidth=2, label=f'Ideal Normal Dist.\n(\u03bc={mu:.1f}, \u03c3={std:.1f})')

    # Draw Mean and Median lines to prove skewness visually
    mean_val = original_data.mean()
    median_val = original_data.median()
    ax1.axvline(mean_val, color='blue', linestyle='dashed', linewidth=2, label=f'Mean: {mean_val:.1f}')
    ax1.axvline(median_val, color='green', linestyle='solid', linewidth=2, label=f'Median: {median_val:.1f}')

    ax1.set_title(f'BEFORE: Original {feature}\nSkewness: {stats.skew(original_data):.2f}', fontsize=13, fontweight='bold')
    ax1.set_ylabel('Density')
    ax1.legend()

    # ==========================================
    # PLOT 2: LOG1P TRANSFORMED (AFTER)
    # ==========================================
    # Apply log1p transformation
    log_data = np.log1p(original_data)

    sns.histplot(log_data, kde=True, stat='density', ax=ax2, color='mediumseagreen', bins=40)

    # Fit a theoretical Normal Distribution curve for the log data
    mu_log, std_log = stats.norm.fit(log_data)
    xmin_log, xmax_log = ax2.get_xlim()
    x_log = np.linspace(xmin_log, xmax_log, 100)
    p_log = stats.norm.pdf(x_log, mu_log, std_log)
    ax2.plot(x_log, p_log, 'k--', linewidth=2, label=f'Ideal Normal Dist.\n(\u03bc={mu_log:.1f}, \u03c3={std_log:.1f})')

    # Draw Mean and Median lines
    mean_log = log_data.mean()
    median_log = log_data.median()
    ax2.axvline(mean_log, color='blue', linestyle='dashed', linewidth=2, label=f'Mean: {mean_log:.1f}')
    ax2.axvline(median_log, color='green', linestyle='solid', linewidth=2, label=f'Median: {median_log:.1f}')

    ax2.set_title(f'AFTER: Log1p Transformed {feature}\nSkewness: {stats.skew(log_data):.2f}', fontsize=13, fontweight='bold')
    ax2.set_ylabel('')
    ax2.legend()

    # Add a main title for the pair
    plt.suptitle(f'Impact of Log Transformation on {feature}', fontsize=16, y=1.05)
    plt.tight_layout()
    plt.show()
```

```python
df_encoded = pd.get_dummies(df_cleaned, drop_first=True)
X = df_encoded.drop('SalePrice', axis=1)
y = np.log1p(df_encoded['SalePrice'])

kf = KFold(n_splits=5, shuffle=True, random_state=42)
```

```python
lasso_model = Lasso(alpha=0.0005, random_state=42, max_iter=10000)
rfr_model = RandomForestRegressor(n_estimators=100, random_state=42)
xgbr_model = XGBRegressor(n_estimators=100, learning_rate=0.1, random_state=42)
```

```python
lasso_cv_result = cross_validate(lasso_model, X, y, cv=kf, scoring=('r2', 'neg_mean_squared_error'), n_jobs=-1)
rfr_cv_result = cross_validate(rfr_model, X, y, cv=kf, scoring=('r2', 'neg_mean_squared_error'), n_jobs=-1)
xgbr_cv_result = cross_validate(xgbr_model, X, y, cv=kf, scoring=('r2', 'neg_mean_squared_error'), n_jobs=-1)
print(lasso_cv_result)
print(rfr_cv_result)
print(xgbr_cv_result)
```

```python
lasso_mean_r2 = np.mean(lasso_cv_result['test_r2'])
lasso_std_r2 = np.std(lasso_cv_result['test_r2'])
lasso_rmse = np.mean(np.sqrt(-lasso_cv_result['test_neg_mean_squared_error']))
print(f"Lasso Regression Model:")
print(f"  Avg RMSE (Log Scale): {lasso_rmse:.4f}")
print(f"  Avg R2 Score:         {lasso_mean_r2:.4f} (+/- {lasso_std_r2 * 2:.4f})\n")
```

```python
rfr_mean_r2 = np.mean(rfr_cv_result['test_r2'])
rfr_std_r2 = np.std(rfr_cv_result['test_r2'])
rfr_rmse = np.mean(np.sqrt(-rfr_cv_result['test_neg_mean_squared_error']))
print(f"Random Forest Regressor Model:")
print(f"  Avg RMSE (Log Scale): {rfr_rmse:.4f}")
print(f"  Avg R2 Score:         {rfr_mean_r2:.4f} (+/- {rfr_std_r2 * 2:.4f})\n")
```

```python
xgbr_mean_r2 = np.mean(xgbr_cv_result['test_r2'])
xgbr_std_r2 = np.std(xgbr_cv_result['test_r2'])
xgbr_rmse = np.mean(np.sqrt(-xgbr_cv_result['test_neg_mean_squared_error']))
print(f"XGradientBoosting Regressor model:")
print(f"  Avg RMSE (Log Scale): {xgbr_rmse:.4f}")
print(f"  Avg R2 Score:         {xgbr_mean_r2:.4f} (+/- {xgbr_std_r2 * 2:.4f})\n")
```

```python
best_model = XGBRegressor(n_estimators=100, learning_rate=0.1, random_state=42)
y_pred_log_cv = cross_val_predict(best_model, X, y, cv=kf, n_jobs=-1)
best_model.fit(X, y)
```

```python
y_actual_price = np.expm1(y)
y_pred_price = np.expm1(y_pred_log_cv)
print(y_actual_price)
print(y_pred_price)
```

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.model_selection import cross_val_predict

def plot_residual_analysis(model, X, y, model_name, cv_object):
    """
    Generates Residual & Error Analysis plots with automatic zoom
    to handle extreme outliers and make the plots readable.
    """
    y_pred_log = cross_val_predict(model, X, y, cv=cv_object, n_jobs=-1)

    y_true_real = np.expm1(y)
    y_pred_real = np.expm1(y_pred_log)
    residuals = y_true_real - y_pred_real

    error_df = pd.DataFrame({'Actual': y_true_real, 'Predicted': y_pred_real, 'Residual': residuals})
    error_df['Error_Type'] = np.where(error_df['Residual'] < 0, 'Overpriced (Pred > Actual)', 'Underpriced (Pred < Actual)')

    over_pct = (error_df['Error_Type'] == 'Overpriced (Pred > Actual)').mean() * 100
    under_pct = 100 - over_pct

    # Calculate limits for zooming in (Ignoring the top/bottom 1% most extreme errors)
    lower_bound = np.percentile(error_df['Residual'], 1)
    upper_bound = np.percentile(error_df['Residual'], 99)
    max_price = np.percentile(error_df['Predicted'], 99)

    fig, axes = plt.subplots(1, 3, figsize=(22, 6))
    fig.suptitle(f'Residual & Error Analysis: {model_name}', fontsize=18, fontweight='bold', y=1.05)

    # Plot 1: Actual vs Predicted Prices
    sns.scatterplot(x='Actual', y='Predicted', data=error_df, alpha=0.5, ax=axes[0], hue='Error_Type', palette={'Overpriced (Pred > Actual)': 'indianred', 'Underpriced (Pred < Actual)': 'mediumseagreen'})
    max_val = max(error_df['Actual'].max(), error_df['Predicted'].max())
    axes[0].plot([0, max_val], [0, max_val], 'k--', lw=2, label='Perfect Prediction (y=x)')
    axes[0].set_title('Actual vs. Predicted Prices', fontsize=12)
    axes[0].set_xlabel('Actual Price ($)')
    axes[0].set_ylabel('Predicted Price ($)')
    axes[0].set_xlim(-10000, max_price * 1.2) # Zoom applied
    axes[0].set_ylim(-10000, max_price * 1.2) # Zoom applied
    axes[0].legend()

    # Plot 2: Residuals vs Predicted
    sns.scatterplot(x='Predicted', y='Residual', data=error_df, alpha=0.5, ax=axes[1], hue='Error_Type', palette={'Overpriced (Pred > Actual)': 'indianred', 'Underpriced (Pred < Actual)': 'mediumseagreen'}, legend=False)
    axes[1].axhline(y=0, color='k', linestyle='--', lw=2)
    axes[1].set_title('Residuals vs. Predicted Prices\n(Zoomed in to ignore extreme 1% outliers)', fontsize=12)
    axes[1].set_xlabel('Predicted Price ($)')
    axes[1].set_ylabel('Residual Error ($)')
    axes[1].set_ylim(lower_bound * 1.2, upper_bound * 1.2) # Zoom applied
    axes[1].set_xlim(-10000, max_price * 1.2) # Zoom applied

    # Plot 3: Distribution of Residuals
    # We filter the data just for the histogram so it draws a proper bell curve
    zoomed_residuals = error_df[(error_df['Residual'] > lower_bound * 1.5) & (error_df['Residual'] < upper_bound * 1.5)]['Residual']
    sns.histplot(zoomed_residuals, kde=True, ax=axes[2], color='steelblue', bins=40)
    axes[2].axvline(x=0, color='k', linestyle='--', lw=2)
    axes[2].set_title(f'Error Distribution (Zoomed)\nOverpriced: {over_pct:.1f}% | Underpriced: {under_pct:.1f}%', fontsize=12)
    axes[2].set_xlabel('Residual Error ($)')
    axes[2].set_ylabel('Frequency')

    plt.tight_layout()
    plt.show()

plot_residual_analysis(xgbr_model, X, y, "XGBoost Regressor", kf)
plot_residual_analysis(lasso_model, X, y, "Lasso Regression", kf)
plot_residual_analysis(rfr_model, X, y, "Random Forest Regressor", kf)
```

```python
import ipywidgets as widgets
from IPython.display import display, clear_output
import pandas as pd
import numpy as np

print("--- Preparing Full Interactive UI ---")

# ==========================================
# 1. DEFINE MAIN AND ADVANCED COLUMNS
# ==========================================
# User defined main features
main_cols = ['Gr Liv Area', 'Overall Cond', 'Overall Qual', 'TotRms AbvGrd', 'Year Built', 'Bldg Type']

# All available columns except the target variable
all_cols = [col for col in df_cleaned.columns if col != 'SalePrice']

# Advanced columns are everything else
adv_cols = [col for col in all_cols if col not in main_cols]

# ==========================================
# 2. DYNAMICALLY GENERATE ALL WIDGETS
# ==========================================
style = {'description_width': '160px'}
layout = widgets.Layout(width='500px')
input_widgets = {}

for col in all_cols:
    # If the column is Categorical (Object) -> Create a Dropdown
    if df_cleaned[col].dtype == 'object':
        options = sorted(list(df_cleaned[col].astype(str).unique()))
        mode_val = df_cleaned[col].mode()[0]

        input_widgets[col] = widgets.Dropdown(
            options=options, value=str(mode_val),
            description=col + ':', style=style, layout=layout
        )

    # If the column is Numerical (Int/Float) -> Create a Slider
    else:
        min_val = df_cleaned[col].min()
        max_val = df_cleaned[col].max()
        median_val = df_cleaned[col].median()

        # Failsafe for constant columns
        if min_val == max_val:
            max_val = min_val + 1

        if pd.api.types.is_float_dtype(df_cleaned[col]):
            input_widgets[col] = widgets.FloatSlider(
                value=median_val, min=min_val, max=max_val,
                description=col + ':', style=style, layout=layout, continuous_update=False
            )
        else:
            input_widgets[col] = widgets.IntSlider(
                value=int(median_val), min=int(min_val), max=int(max_val),
                description=col + ':', style=style, layout=layout, continuous_update=False
            )

# ==========================================
# 3. ORGANIZE UI LAYOUT
# ==========================================
# Build Main UI Box
main_boxes = [input_widgets[col] for col in main_cols]
main_ui = widgets.VBox([widgets.HTML("<h3>🏠 Interactive House Price Estimator</h3>")] + main_boxes)

# Build Advanced UI Box (with scrollable y-axis to save space)
adv_boxes = [input_widgets[col] for col in adv_cols]
adv_vbox = widgets.VBox(adv_boxes, layout=widgets.Layout(max_height='350px', overflow_y='auto'))

accordion = widgets.Accordion(children=[adv_vbox])
accordion.set_title(0, f'⚙️ Advanced / Optional Features ({len(adv_cols)} fields)')
accordion.selected_index = None # Closed by default

# ==========================================
# 4. PREDICTION LOGIC & INFERENCE
# ==========================================
predict_btn = widgets.Button(description='Predict Price', button_style='success', icon='calculator')
output_area = widgets.Output()

def on_predict_clicked(b):
    with output_area:
        clear_output()
        print("Calculating dynamically...")

        # 1. Gather all current values from widgets
        user_data = {col: input_widgets[col].value for col in all_cols}
        user_df = pd.DataFrame([user_data])

        # 2. Apply the EXACT same encoding used in training
        # This converts categorical dropdowns back to dummy (0/1) columns
        user_encoded = pd.get_dummies(user_df, drop_first=True, dtype=int)

        # 3. Reindex to align exactly with the training features (X.columns)
        # Missing dummy columns (categories not selected) are automatically filled with 0
        user_encoded = user_encoded.reindex(columns=X.columns, fill_value=0)

        # 4. Predict and convert from Log to Dollar
        log_price = best_model.predict(user_encoded)[0]
        actual_price = np.expm1(log_price)

        clear_output()
        print("="*50)
        print(f"💰 ESTIMATED SALE PRICE: ${actual_price:,.2f}")
        print("="*50)

predict_btn.on_click(on_predict_clicked)

# ==========================================
# 5. DISPLAY THE FINAL WIDGET
# ==========================================
final_ui = widgets.VBox([
    main_ui,
    widgets.HTML("<hr>"),
    accordion,
    widgets.HTML("<br>"),
    predict_btn,
    output_area
])

display(final_ui)
```

```python
"""
## 1. Feature Selection Impacts
The handling of missing data was a critical component of our feature engineering strategy. Instead of dropping columns with high null
counts (e.g., `Pool QC`, `Garage Type`), we encoded them as "None" or `0`. This strategy allowed the model to use "missingness" as a meaningful feature.
For instance, the absence of a garage or a basement natively applied a penalty to the predicted price.
Furthermore, L1 Regularization (Lasso) was utilized during benchmarking to validate feature importance. Lasso effectively reduced the coefficients of
noisy or irrelevant variables to zero, confirming that features like `Overall Qual`, `Gr Liv Area`, and `Total Bsmt SF` hold the highest predictive
power.

## 2. Overfitting vs. Generalization
To prevent the model from memorizing the training data (overfitting), we strictly implemented a 5-Fold Cross-Validation strategy.
* **Overfitting Prevention:** By training and evaluating the models across 5 distinct subsets of the data, we ensured that our evaluation metrics were
robust and not the result of a lucky train-test split.
* **Generalization Performance:** The champion model, XGBoost, achieved a consistent Cross-Validated R2 score of ~0.90. This high out-of-fold
performance proves that the model generalizes exceptionally well to unseen data.

## 3. Error & Residual Analysis
Our residual analysis (using `cross_val_predict`) demonstrated a normal distribution of errors centered around zero. However, a slight right-skewness
indicated a minor tendency toward *underpricing* in the ultra-luxury segment. The model performs exceptionally well for typical homes but behaves
conservatively when estimating extreme outlier properties with rare premium features.
"""
```
